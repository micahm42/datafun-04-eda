## Spotify Data Observations

The dataset contains information about songs, including
popularity, energy, danceability, and streams.

One observation is that songs with higher popularity
may have a positive relationship with the number of streams.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load the Spotify dataset
df = pd.read_csv(r"C:\Repos\datafun-04-eda\data\spotify.csv")

# Display the first five rows
df.head()

sns.set_theme(style="whitegrid", palette="Set2")

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Popularity",
    bins=20,
    kde=True,
    color="teal"
)

plt.title("Distribution of Spotify Track Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Tracks")

plt.show()
```

```python

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Genre",
    y="Popularity",
    color="skyblue"
)

plt.title("Popularity Distribution by Genre")
plt.xlabel("Genre")
plt.ylabel("Popularity")

plt.xticks(rotation=45)

plt.show()
```

```python
# Filter the dataset to highly popular tracks

df_subset = df[df["Popularity"] >= 70]

print("Original dataset shape:", df.shape)
print("Filtered dataset shape:", df_subset.shape)

df_subset.head()
```

```python

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df_subset,
    x="Streams (M)",
    bins=15,
    kde=True,
    color="coral"
)

plt.title("Streams Distribution for Popular Tracks")
plt.xlabel("Streams (Millions)")
plt.ylabel("Number of Tracks")

plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df_tips, x='total_bill', bins=20, kde=True, color='teal')
plt.title('Distribution of Total Bills')
plt.xlabel('Total Bill ($)')
plt.ylabel('Number of Tables')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_tips, x='total_bill', y='tip', hue='time', style='sex', s=80)
plt.title('Total Bill vs. Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_tips, x='day', y='tip', color='lightblue')
plt.title('Tip Distribution by Day')
plt.xlabel('Day')
plt.ylabel('Tip ($)')
plt.show()
```
